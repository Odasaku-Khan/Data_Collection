```python
female_passengers = df[df["Sex"] == "female"]
female_passengers[["Name","Sex","Age","Survived"]]
```

```python
first_class_female_survivors = df[
    (df["Sex"] == "female") &
    (df["Pclass"] == 1) &
    (df["Survived"] == 1)
]

first_class_female_survivors[
    ["Name", "Sex", "Age", "Pclass", "Survived"]
]
```

```python
average_fare = df["Fare"].mean()

average_fare
```
